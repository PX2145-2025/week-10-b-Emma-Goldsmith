Exercise 1 -  Plot the original signal x against time instead of point number. (Use N=64 points)

```python
# Examine a small number of points to see what happens more easily
N = 64 # the fft algorithm is fastest for N a power of 2
tau = 10.0 # time length
t = np.linspace(0, tau, N, endpoint=False) # creates N points to cover tau seconds
# Choose endpoint False so that signal would fit exactly periodically within tau
f = 0.2 # frequency in Hz
x = np.cos(2*np.pi*f*t) # create x array, amplitude 1.0
plt.plot(t, x)
plt.plot(t, x,'o') # show as points and line
# Note how the plot is exactly periodic

plt.title('Single cosine wave sampled at {} points'.format(N))
plt.xlabel('time')
plt.ylabel('x(N)')
plt.grid()
```

Exercise 2 - Plot just the positive frequencies of the FFT using the fft routine of numpy. This is a plot versus point number.

```python
X = fft(x)/N # fft and normalise
plt.plot(np.abs(X)[0:32], 'o') # first half is poitivie
plt.title('positive frequenies of the FFT sampled at '+str(N)+' points')
plt.xlabel('point number')
plt.ylabel('$\mid X(N)\mid$')
plt.grid()
```

Exercise 3 -  For this question and all the following, use the Spectrum1 function. Calculate and plot the FFT of x versus frequency. (In the rest of the exercises just plot against frequency.) Why is the signal amplitude you find equal to 0.5 instead of 1?

Recall: The sampling rate is
$F_S = \frac{1}{Δt}= \frac{N}{τ}$
. Read the section Summary above!

```python
def Spectrum1(x, Fs):
    '''
    One-sided amplitude spectrum of x(t).
    '''
    N = len(x) # length of the signal
    frq = (float(Fs)/N)*np.arange(0, N/2+1) # one-sided frequency range
    X = fft(x)/N # fft computing and normalization
    X = X[0:N//2+1] # take just positive values
    # (in python3, // ensures integer division)

    # frq is array of actual frequencies
    # X is complex array of amplitudes
    return frq , X
```

```python
#F_s is sample rate = N/tau
Fs = N/tau

frq_result, X_result = Spectrum1(x, Fs)

plt.figure()
plt.plot(frq_result, np.abs(X_result))
plt.plot(frq_result, np.abs(X_result), 'o')
plt.xlabel('Frequency')
plt.ylabel('Magnitude of X(N)')
plt.title('FFT of x vs frequency')
plt.grid()
plt.show()
```

The signal amplitude is equal to 0.5 instead of 1 because it has been normalised.

```python
N = 64 # the fft algorithm is fastest for N a power of 2
tau = 10.0 # time length
t = np.linspace(0, tau, N, endpoint=False) # creates N points to cover tau seconds
# Choose endpoint False so that signal would fit exactly periodically within tau
f = 0.2 # frequency in Hz
x = np.cos(2*np.pi*f*t) # create x array, amplitude 1.0


f_2 = 0.22 # frequency in Hz
x_2 = np.cos(2*np.pi*f_22*t) # create x array, amplitude 1.0

plt.plot(t, x, label='frequency = 0.2')
plt.plot(t, x,'o') # show as points and line
plt.plot(t, x_2, 'r--', label='frequency = 0.22')
plt.plot(t, x_2,'o') # show as points and line
# Note how the plot is exactly periodic
plt.title('Single cosine wave sampled at {} points'.format(N))
plt.xlabel('time')
plt.ylabel('x(N)')
plt.legend()
plt.grid()
plt.show()
```

```python
N = 64 # the fft algorithm is fastest for N a power of 2
tau = 10.0 # time length
t = np.linspace(0, tau, N, endpoint=False) # creates N points to cover tau seconds
f = 0.2 # frequency in Hz
x = np.cos(2*np.pi*f*t) # create x array, amplitude 1.0

N_2 = 150 # the fft algorithm is fastest for N a power of 2
t_2 = np.linspace(0, tau, N_2, endpoint=False)
x_2 = np.cos(2*np.pi*f*t_2) # create x array, amplitude 1.0

N_3 = 250 # the fft algorithm is fastest for N a power of 2
t_3 = np.linspace(0, tau, N_3, endpoint=False)
x_3 = np.cos(2*np.pi*f*t_3) # create x array, amplitude 1.0

plt.plot(t, x, label='number of points = 64')
plt.plot(t_2, x_2,'g--', label='number of points = 150')
plt.plot(t_3, x_3,'r--', label='number of points = 250')


plt.title('Single cosine wave sampled at ranging points')
plt.xlabel('time')
plt.ylabel('x(N)')
plt.legend()
plt.grid()
plt.show()
```

b) what happens as you increase the lenght τ of the original data (for both 0.2 and 0.22 Hz).

```python
N = 64 # the fft algorithm is fastest for N a power of 2

f = 0.2 # frequency in Hz
f_2 = 0.22

tau = 10.0 # time length
t = np.linspace(0, tau, N, endpoint=False)

tau_2 = 25.0 # time length
t_2 = np.linspace(0, tau_2, N, endpoint=False)

x = np.cos(2*np.pi*f*t)
x_2 = np.cos(2*np.pi*f_2*t)
x_3 = np.cos(2*np.pi*f*t_2)
x_4 = np.cos(2*np.pi*f_2*t_2)

plt.plot(t, x, label='tau = 10, frequency = 0.2')
plt.plot(t, x_2, 'r--', label='tau = 10, frequency = 0.22')
plt.plot(t_2, x_3, label='tau = 25, frequency = 0.2')
plt.plot(t_2, x_4,'g--', label='tau = 25, frequency = 0.22')

plt.title('Single cosine wave sampled at ranging frequencies and values of tau')
plt.xlabel('time')
plt.ylabel('x(N)')
plt.legend()
plt.grid()
plt.show()
```
